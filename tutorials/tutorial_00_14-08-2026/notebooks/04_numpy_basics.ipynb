## 1. Creating arrays

`np.array` turns a Python list into a fast, math-aware array.
An array knows its **shape** (how many elements per dimension) and its **dtype** (the type of the numbers).

````python
# Build a 1D array from a list
a = np.array([1, 2, 3, 4, 5])
print("a =", a)
print("shape:", a.shape)     # (5,)  -> 5 elements, one dimension
print("dtype:", a.dtype)     # int64 -> integer elements

# Indexing and slicing work exactly like lists
print("first:", a[0])        # 1
print("last :", a[-1])       # 5
print("slice:", a[1:4])      # elements at positions 1, 2, 3

# A 2D array (a matrix): shape = (rows, columns)
m = np.array([[1, 2], [3, 4]])
print("2D array:\n", m)
print("2D shape:", m.shape)  # (2, 2)
print("row 0, col 1:", m[0, 1])   # 2
````

## 2. The big five array creators

> 🎯 **`np.linspace` is the most-used NumPy function in this course** — it
> generates the x-values for every plot you will make.

````python
# np.arange(start, stop, step) -> evenly spaced INTEGERS
print("arange:", np.arange(0, 10, 2))      # [0 2 4 6 8]

# np.linspace(a, b, n) -> n evenly spaced values from a to b (INCLUSIVE)
x = np.linspace(0, 1, 5)
print("linspace:", x)                       # [0. 0.25 0.5 0.75 1.]

# np.zeros / np.ones -> arrays filled with 0 or 1
print("zeros(2,3):\n", np.zeros((2, 3)))

# np.zeros_like(x) -> zeros with the SAME shape as x
zero_grid = np.zeros_like(x)
print("zeros_like(x):", zero_grid)          # same length as x, all zeros
````

## 3. Random numbers: `np.random`

Simulations are full of randomness (Monte Carlo!). NumPy gives us every
distribution we need.

> 🎯 **Always set the seed first** (`np.random.seed(...)`) so your "random"
> results can be reproduced — the heart of scientific computing.

````python
# ALWAYS set the seed first so results are reproducible
np.random.seed(42)

print("rand(5)      :", np.random.rand(5))            # uniform in [0, 1)
print("randint(0,10):", np.random.randint(0, 10, 5))  # random integers
print("normal(0,1)  :", np.random.normal(0, 1, 5))    # Gaussian
print("uniform(-1,1):", np.random.uniform(-1, 1, 5))  # uniform in [-1, 1)

# A 3x3 grid of random positions (like atom coordinates!)
positions = np.random.rand(3, 3)
print("random positions (3x3):\n", positions)
````

## 4. Angles: degrees ↔ radians

Simulation code works in **radians**, but humans think in **degrees**.
`np.radians` converts; `np.degrees` converts back. Trig functions
(`np.sin`, `np.cos`, ...) always expect **radians**.

````python
deg = 90
rad = np.radians(deg)          # 90 degrees = pi/2 radians
print("radians(90) =", rad)    # ~1.5708
print("degrees(rad) =", np.degrees(rad))   # back to 90.0

# Trig functions expect radians — always convert first!
angles = np.linspace(0, 360, 37)          # 37 angles in degrees
cos_vals = np.cos(np.radians(angles))     # cos needs radians
print("cos at 0, 90, 180 deg:", cos_vals[0], cos_vals[9], cos_vals[18])
````

## 5. Element-wise math — no loops needed!

NumPy applies math to **every element at once**. Here is the Lennard-Jones
potential — the exact pattern you will use in the simulation sessions:
generate a range of distances with `linspace`, then apply the formula to the
whole array.

````python
# Lennard-Jones potential: E(r) = 4*eps*((sigma/r)**12 - (sigma/r)**6)
r = np.linspace(0.3, 1.0, 50)          # 50 distances in nm
sigma = 0.34                           # LJ sigma (nm)
epsilon = 0.997                        # LJ epsilon (kJ/mol)

sr = sigma / r                         # divide EVERY element by r
lj = 4 * epsilon * (sr**12 - sr**6)    # apply the formula to the whole array

print("LJ energy: min =", round(lj.min(), 4), " max =", round(lj.max(), 4))
print("mean:", round(lj.mean(), 4), " sum:", round(lj.sum(), 4))
````
